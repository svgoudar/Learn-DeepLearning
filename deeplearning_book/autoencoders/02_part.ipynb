```
Input  →  Encoder  →  Latent Space  →  Decoder  →  Output (Reconstruction)
```

| Component                     | Function                                                    |
| ----------------------------- | ----------------------------------------------------------- |
| **Encoder**                   | Compresses input into lower-dimensional representation.     |
| **Latent space (bottleneck)** | Holds the essential information.                            |
| **Decoder**                   | Reconstructs the original input from the compressed vector. |

---

### Workflow Steps

#### **Step 1 — Input Preparation**

You start with raw input data $x$, often normalized between 0 and 1.
For MNIST: each image (28×28) is flattened into a vector of size **784**.

$$
x \in \mathbb{R}^{784}
$$

---

#### **Step 2 — Encoding (Compression)**

The encoder learns to map input $x$ → latent representation $z$:

$$
z = f_{\text{encoder}}(x)
$$

This is done using a series of **linear layers** + **non-linear activations** (ReLU, tanh, etc.).
The output $z$ has **fewer dimensions** than $x$.
Example: $784 \to 128 \to 64 \to 32$.

Intuitively, the encoder learns **important patterns** and ignores noise or redundancy.

---

#### **Step 3 — Bottleneck / Latent Representation**

This is the **compressed version** of the data:

* Acts as a **low-dimensional code**.
* Ideally captures only the **most informative features**.
* Used later for visualization or as input to other models.

If $z \in \mathbb{R}^{32}$, it means each image is now described by only 32 learned features.

---

#### **Step 4 — Decoding (Reconstruction)**

The decoder tries to reconstruct $x'$ from $z$:

$$
\hat{x} = f_{\text{decoder}}(z)
$$

Structure mirrors the encoder but expands the dimensions back: $32 \to 64 \to 128 \to 784$.

The decoder learns how to **recreate the original input** from compressed information.

---

#### **Step 5 — Reconstruction Loss**

The network is trained to minimize the difference between the original input $x$ and reconstructed output $\hat{x}$.

Common loss functions:

* **Mean Squared Error (MSE)** for continuous data
  $$
  L = |x - \hat{x}|^2
  $$
* **Binary Cross-Entropy (BCE)** for normalized pixel data
  $$
  L = -[x \log(\hat{x}) + (1-x)\log(1-\hat{x})]
  $$

This forces the encoder-decoder pair to **preserve essential information**.

---

#### **Step 6 — Backpropagation and Optimization**

Use optimizers (e.g., Adam, SGD) to update weights of encoder and decoder jointly.

$$
\theta_{\text{new}} = \theta_{\text{old}} - \eta \frac{\partial L}{\partial \theta}
$$

Goal: minimize reconstruction error over all training samples.

---

#### **Step 7 — Inference (Usage)**

Once trained:

* The **encoder** alone can be used for **feature extraction**.
* The **decoder** can be used for **data generation** (given latent vectors).
* Combined, the full autoencoder can perform:

  * **Noise removal (denoising AE)**
  * **Anomaly detection (reconstruction error high → anomaly)**
  * **Dimensionality reduction (latent space)**

---

### **Workflow Summary Diagram**

```

```

---

**Example Use Cases**

| Application                       | Description                                                     |
| --------------------------------- | --------------------------------------------------------------- |
| **Denoising Autoencoder**         | Learns to remove noise from inputs.                             |
| **Sparse Autoencoder**            | Forces some neurons to stay inactive → compact representations. |
| **Variational Autoencoder (VAE)** | Learns probabilistic latent space; can generate new data.       |
| **Contractive Autoencoder**       | Adds penalty to make encoding robust to input changes.          |
| **Deep Autoencoder**              | Uses many hidden layers for high-level features.                |
