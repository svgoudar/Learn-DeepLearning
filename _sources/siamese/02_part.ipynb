
## Workflow

A **Siamese Network** learns to measure *similarity* between two inputs through a structured and repeated workflow.
It follows a clear sequence from data preparation to learning an embedding space where *closeness = similarity* and *distance = difference.*

---

### **1. Input Preparation**

**Goal:** Create pairs of data to train the model to distinguish “same” vs “different.”

* **Positive pairs:** Two inputs from the same class.
  Example → (Face of Person A, Face of Person A)
* **Negative pairs:** Two inputs from different classes.
  Example → (Face of Person A, Face of Person B)

**Labels:**

* 1 → Same class (similar)
* 0 → Different class (dissimilar)

> The dataset now becomes *pair-based* rather than *instance-based.*

---

### **2. Feature Extraction (Twin Networks)**

Each input in a pair passes through **identical subnetworks** that share the same weights.
These networks are usually CNNs (for images), LSTMs (for text), or feedforward layers (for tabular data).

$$
h_1 = f(x_1; \theta), \quad h_2 = f(x_2; \theta)
$$

Here:

* $f(\cdot)$ → shared neural network
* $h_1, h_2$ → embeddings or feature vectors

**Purpose:** Encode both inputs into a *common representation space.*

---

### **3. Distance Computation**

A **distance metric** measures how far apart the two embeddings are.
Common choices:

* **Euclidean distance:**
  $$
  D = |h_1 - h_2|_2
  $$
* **Cosine similarity:**
  $$
  D = 1 - \frac{h_1 \cdot h_2}{|h_1||h_2|}
  $$

If two inputs are similar → $D$ is small.
If different → $D$ is large.

---

### **4. Loss Calculation**

The model uses the **Contrastive Loss** to adjust its weights.

$$
L = (1 - y)\frac{1}{2}D^2 + (y)\frac{1}{2}\max(0, m - D)^2
$$

Where:

* $y = 0$: similar pair → distance should be small.
* $y = 1$: dissimilar pair → distance should be larger than margin $m$.
* $m$: safety margin defining separation between similar and dissimilar samples.

**Intuition:**

* Pull embeddings of similar inputs closer.
* Push embeddings of dissimilar inputs apart.

---

### **5. Backpropagation and Weight Update**

Gradients are computed based on the loss and propagated **through both subnetworks simultaneously** (since they share weights).
This ensures both networks evolve together to learn meaningful representations.

---

### **6. Inference / Testing Phase**

During testing:

1. Take two new inputs (never seen before).
2. Pass them through the trained Siamese model to get embeddings.
3. Compute distance $D$.
4. Compare with a threshold:

   * $D < \text{threshold}$ → similar.
   * $D > \text{threshold}$ → dissimilar.

> The model can now recognize similarity *without ever being explicitly trained on that specific class.*

---

### **7. Embedding Visualization (Optional)**

You can visualize embeddings (e.g., with t-SNE) to see clusters:

* Similar items cluster together.
* Dissimilar ones scatter apart.

This shows that the model has successfully learned a **semantic similarity space.**

---

### **🧩 Workflow Summary**

| Step | Process                    | Purpose                                         |
| ---- | -------------------------- | ----------------------------------------------- |
| 1    | Prepare pairs              | Provide data to learn similarity                |
| 2    | Shared network encoding    | Create consistent embeddings                    |
| 3    | Compute distance           | Quantify similarity                             |
| 4    | Contrastive / Triplet loss | Pull similar pairs, push apart dissimilar pairs |
| 5    | Backpropagate              | Update shared weights                           |
| 6    | Infer on new data          | Compare unseen pairs                            |
| 7    | Visualize embeddings       | Verify learned structure                        |

---

**End result:**
A Siamese Network learns a **function that transforms any input into a numerical fingerprint** — where closeness in space directly represents similarity in meaning or appearance.
