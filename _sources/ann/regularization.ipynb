```
Data → Model → Loss + Regularization → Backprop → Update → Validation
```

**Practical steps**

1. Choose base model
2. Add explicit regularizers (L1/L2, Dropout)
3. Train while monitoring validation loss
4. Apply early stopping
5. Tune λ via validation

---

### Remediation: Detecting & Fixing Overfitting

| Symptom                       | Diagnosis     | Remedy                  |
| ----------------------------- | ------------- | ----------------------- |
| Low train loss, high val loss | Overfitting   | Increase regularization |
| Both losses high              | Underfitting  | Reduce regularization   |
| Unstable training             | High variance | Add dropout / BN        |

---

### PyTorch Demonstration

#### Baseline Model

```python
import torch
import torch.nn as nn

class Net(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(20, 128),
            nn.ReLU(),
            nn.Linear(128, 1)
        )

    def forward(self, x):
        return self.net(x)
```

#### L2 Regularization (Weight Decay)

```python
model = Net()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3, weight_decay=1e-4)
```

`weight_decay` implements L2 penalty.

---

### Summary

| Aspect             | Insight                                 |
| ------------------ | --------------------------------------- |
| Purpose            | Improve generalization                  |
| Core principle     | Penalize complexity                     |
| Where applied      | Loss, architecture, data, training loop |
| Key hyperparameter | Regularization strength λ               |

Regularization is not a single method but a **design philosophy** that shapes model behavior toward simplicity, robustness, and reliability.
