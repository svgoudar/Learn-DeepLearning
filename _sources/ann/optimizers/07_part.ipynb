## RMSProp Optimizer in Deep Learning

### Concept Overview

**RMSProp (Root Mean Square Propagation)** is an adaptive learning rate optimization algorithm designed to address the limitations of vanilla Gradient Descent and Adagrad, especially in training deep neural networks with **non-stationary objectives** and **ill-conditioned curvature**.

It stabilizes training by **normalizing gradients using a moving average of their recent magnitudes**, preventing both exploding and vanishing updates.

---

### Motivation

Standard SGD uses a single learning rate:

$$
\theta_{t+1} = \theta_t - \eta \nabla \mathcal{L}(\theta_t)
$$

Problems:

* Large gradients → unstable updates
* Small gradients → slow convergence
* Different parameters need different step sizes

RMSProp introduces **per-parameter adaptive learning rates**.

---

### Mathematical Formulation

For each parameter $\theta$:

$$
g_t = \nabla_\theta \mathcal{L}_t
$$

$$
s_t = \rho s_{t-1} + (1 - \rho) g_t^2
$$

$$
\theta_{t+1} = \theta_t - \frac{\eta}{\sqrt{s_t + \epsilon}} g_t
$$

Where:

| Symbol       | Meaning                                                    |
| ------------ | ---------------------------------------------------------- |
| $g_t$      | Current gradient                                           |
| $s_t$      | Exponentially weighted moving average of squared gradients |
| $\rho$     | Decay rate (typically 0.9)                                 |
| $\eta$     | Learning rate                                              |
| $\epsilon$ | Numerical stability constant                               |

---

### Intuition

RMSProp **divides each gradient by its recent RMS value**:

* Large historical gradients → smaller steps
* Small historical gradients → larger steps

This produces:

* Stable learning
* Faster convergence
* Robustness to poorly scaled problems

---

### Training Workflow Impact

| Phase            | Effect                |
| ---------------- | --------------------- |
| Backpropagation  | Compute raw gradients |
| RMS accumulation | Track gradient energy |
| Normalization    | Scale updates         |
| Parameter update | Stable descent        |

---

### PyTorch Demonstration

```python
import torch
import torch.nn as nn
import torch.optim as optim

model = nn.Sequential(
    nn.Linear(2, 64),
    nn.ReLU(),
    nn.Linear(64, 2)
)

optimizer = optim.RMSprop(model.parameters(), lr=0.01, alpha=0.9, eps=1e-8)
criterion = nn.CrossEntropyLoss()

x = torch.randn(128, 2)
y = torch.randint(0, 2, (128,))

for _ in range(100):
    optimizer.zero_grad()
    loss = criterion(model(x), y)
    loss.backward()
    optimizer.step()
```

---

### Key Hyperparameters

| Parameter  | Description         | Typical Value |
| ---------- | ------------------- | ------------- |
| `lr`       | Base learning rate  | 1e-3 to 1e-2  |
| `alpha`    | Decay rate $\rho$ | 0.9           |
| `eps`      | Stability constant  | 1e-8          |
| `momentum` | Optional momentum   | 0.0–0.9       |

---

### Comparison with Other Optimizers

| Optimizer | Main Idea                                       |
| --------- | ----------------------------------------------- |
| SGD       | Fixed learning rate                             |
| Adagrad   | Accumulates all past gradients                  |
| RMSProp   | Exponential moving average of squared gradients |
| Adam      | RMSProp + Momentum bias correction              |

---

### Variants

| Variant            | Modification                          |
| ------------------ | ------------------------------------- |
| RMSProp + Momentum | Adds velocity term                    |
| Centered RMSProp   | Normalizes by variance instead of RMS |
| AdaDelta           | Learning rate-free variant            |

---

### When to Use RMSProp

Use RMSProp when:

* Training deep networks with unstable gradients
* Handling non-stationary objectives (e.g., RNNs, RL)
* SGD converges slowly or oscillates
* You need robust performance without heavy tuning

---

### Summary

| Property                | Benefit                     |
| ----------------------- | --------------------------- |
| Adaptive learning rates | Faster convergence          |
| Gradient normalization  | Stability                   |
| Exponential memory      | Responsive to recent trends |
| Low tuning overhead     | Practical optimizer         |

---

### Final Insight

RMSProp is one of the first optimizers to make **deep learning stable and practical** by explicitly controlling the scale of parameter updates, forming the foundation of modern adaptive methods such as **Adam**.
