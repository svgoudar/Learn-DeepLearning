## Mini-Batch Gradient Descent Optimizer

### Concept Overview

**Mini-batch Gradient Descent (MBGD)** is the most widely used optimization strategy for training deep neural networks.
It updates model parameters using the average gradient computed over a **small random subset of training data (mini-batch)** rather than the entire dataset (batch) or a single sample (stochastic).

---

### Why Mini-Batch?

| Method            | Update Frequency | Stability    | Speed             |
| ----------------- | ---------------- | ------------ | ----------------- |
| Batch GD          | Once per epoch   | Very stable  | Very slow         |
| Stochastic GD     | Every sample     | Very noisy   | Fast but unstable |
| **Mini-Batch GD** | Every batch      | **Balanced** | **Fast & stable** |

Mini-batch GD achieves an optimal trade-off between **computational efficiency** and **training stability**.

---

### Mathematical Formulation

For a mini-batch $B \subset D$:

$$
\theta \leftarrow \theta - \eta \cdot \frac{1}{|B|} \sum_{(x,y) \in B} \nabla_\theta \mathcal{L}(x, y; \theta)
$$

Where:

* $\theta$ = model parameters
* $\eta$ = learning rate
* $\mathcal{L}$ = loss function

---

### Intuition

Mini-batches provide:

* **Noisy but informative gradients**
* Faster convergence than batch GD
* More stable convergence than SGD
* Efficient GPU utilization

Noise introduced by mini-batches helps the optimizer **escape poor local minima and saddle points**.

---

### Training Workflow

1. Shuffle training data
2. Split into mini-batches
3. For each mini-batch:

   * Forward pass
   * Compute loss
   * Backpropagate gradients
   * Update parameters
4. Repeat for multiple epochs

---

### PyTorch Demonstration

#### Dataset



````python
from sklearn.datasets import make_classification
from torch.utils.data import DataLoader, TensorDataset
import torch
import torch.nn as nn

X, y = make_classification(n_samples=3000, n_features=20)
X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.long)

loader = DataLoader(TensorDataset(X, y), batch_size=64, shuffle=True)
````



#### Model



````python
model = nn.Sequential(
    nn.Linear(20, 64),
    nn.ReLU(),
    nn.Linear(64, 2)
)
````



#### Training with Mini-Batch Gradient Descent



````python
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)
loss_fn = nn.CrossEntropyLoss()

for epoch in range(10):
    for xb, yb in loader:
        optimizer.zero_grad()
        pred = model(xb)
        loss = loss_fn(pred, yb)
        loss.backward()
        optimizer.step()
````



Each loop iteration updates the model using a **mini-batch**.

---

### Effect of Batch Size

| Batch Size       | Behavior                               |
| ---------------- | -------------------------------------- |
| Small (16–64)    | Noisy gradients, better generalization |
| Medium (128–512) | Balanced performance                   |
| Large (>1024)    | Stable but may generalize poorly       |

---

### Variants & Extensions

| Variant  | Key Idea                     |
| -------- | ---------------------------- |
| Momentum | Accelerates convergence      |
| Nesterov | Look-ahead momentum          |
| AdaGrad  | Per-parameter learning rates |
| RMSProp  | Gradient normalization       |
| Adam     | Momentum + adaptive rates    |

All modern optimizers are **mini-batch based**.

---

### Practical Guidelines

* Typical batch size: **32–256**
* Always shuffle data each epoch
* Tune learning rate jointly with batch size
* Larger batches require larger learning rates
