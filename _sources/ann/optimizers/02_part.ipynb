
## Stochastic Gradient Descent (SGD) Optimizer in Deep Learning

### Concept Overview

**Stochastic Gradient Descent (SGD)** is the fundamental optimization algorithm used to train neural networks.
It updates model parameters by moving them in the direction that **minimizes the loss function**, using **small random subsets of data** (mini-batches).

Unlike full-batch gradient descent, SGD introduces controlled randomness, which improves convergence speed and generalization.

---

### Mathematical Foundation

Given model parameters $\theta$ and loss $\mathcal{L}(\theta; x, y)$, the update rule is:

$$
\theta_{t+1} = \theta_t - \eta \cdot \nabla_\theta \mathcal{L}(\theta_t; x_i, y_i)
$$

Where:

| Symbol                 | Meaning              |
| ---------------------- | -------------------- |
| $\eta$               | Learning rate        |
| $\nabla \mathcal{L}$ | Gradient of loss     |
| $x_i, y_i$           | Sample or mini-batch |

---

### Intuition

| Property           | Explanation                                     |
| ------------------ | ----------------------------------------------- |
| Random sampling    | Introduces noise that helps escape local minima |
| Frequent updates   | Faster convergence than batch GD                |
| Noise in gradients | Improves generalization                         |
| Memory efficient   | Scales to large datasets                        |

---

### Training Workflow

1. Initialize model parameters
2. Sample mini-batch from dataset
3. Compute predictions
4. Compute loss
5. Backpropagate gradients
6. Update parameters
7. Repeat until convergence

---

### Variants of Gradient Descent

| Variant        | Update Data               |
| -------------- | ------------------------- |
| Batch GD       | Entire dataset            |
| SGD            | Single sample             |
| Mini-batch SGD | Small batch (most common) |

---

### Core Problems with Vanilla SGD

| Issue                      | Effect           |
| -------------------------- | ---------------- |
| Slow convergence           | Zig-zag behavior |
| Sensitive to learning rate | Can diverge      |
| Gets stuck in plateaus     | Slow progress    |

These led to enhanced variants such as **Momentum, RMSProp, Adam**.

---

### PyTorch Demonstration

#### Dataset

```python
from sklearn.datasets import make_moons
import torch
import torch.nn as nn

X, y = make_moons(n_samples=2000, noise=0.2)
X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.long)
```

---

#### Model

```python
model = nn.Sequential(
    nn.Linear(2, 32),
    nn.ReLU(),
    nn.Linear(32, 2)
)
```

---

#### Training with SGD

```python
criterion = nn.CrossEntropyLoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)

for epoch in range(50):
    optimizer.zero_grad()
    output = model(X)
    loss = criterion(output, y)
    loss.backward()
    optimizer.step()
```

---

### Momentum: SGD Variant

**Momentum** accelerates SGD by accumulating past gradients:

$$
v_t = \gamma v_{t-1} + \eta \nabla \mathcal{L}
$$
$$
\theta_{t+1} = \theta_t - v_t
$$

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.1, momentum=0.9)
```

### Comparison with Modern Optimizers

| Optimizer | Strength                |
| --------- | ----------------------- |
| SGD       | Strong generalization   |
| Adam      | Fast convergence        |
| RMSProp   | Handles noisy gradients |
| Adagrad   | Sparse data             |

---

### When to Use SGD

Use SGD when:

* Training very large models
* You need strong generalization
* You want stable long-term convergence

Use Adam when:

* Rapid convergence is required
* Dataset is noisy or sparse

---

### Key Takeaway

> **SGD is the foundation of deep learning optimization; modern optimizers are engineered improvements on its principles.**
