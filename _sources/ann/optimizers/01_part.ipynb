## Gradient Descent Optimizer in Deep Learning

**Gradient Descent (GD)** is the fundamental optimization algorithm used to train neural networks.
Its objective is to **minimize the loss function** by iteratively adjusting model parameters in the direction of the **steepest decrease** of the loss.

$$
\theta_{t+1} = \theta_t - \eta \nabla_\theta \mathcal{L}(\theta_t)
$$

where:

* $\theta$ = model parameters
* $\eta$ = learning rate
* $\nabla_\theta \mathcal{L}$ = gradient of loss w.r.t parameters

---

### Intuition

Imagine standing on a mountain in fog and trying to reach the lowest valley:

* The **slope beneath your feet** is the gradient
* You step **downhill** proportional to how steep it is
* The learning rate controls **step size**

Too large → you overshoot
Too small → you converge very slowly

---

### Training Workflow with Gradient Descent

| Stage            | Operation              |
| ---------------- | ---------------------- |
| Forward pass     | Compute predictions    |
| Loss computation | Measure error          |
| Backward pass    | Compute gradients      |
| Parameter update | Apply gradient descent |
| Repeat           | Until convergence      |

---

### Variants of Gradient Descent

| Variant             | Update Data   | Properties         |
| ------------------- | ------------- | ------------------ |
| Batch GD            | Full dataset  | Stable but slow    |
| Stochastic GD (SGD) | One sample    | Fast but noisy     |
| Mini-batch GD       | Small batches | Practical standard |

---

### Convergence Behavior

| Property                | Effect                      |
| ----------------------- | --------------------------- |
| Convex loss             | Guaranteed global minimum   |
| Non-convex loss         | Local minima, saddle points |
| Ill-conditioned surface | Slow convergence            |

---

### PyTorch Demonstration

#### Simple Linear Regression with SGD



````python
import torch
import torch.nn as nn

X = torch.randn(100, 1)
y = 3 * X + 2 + 0.1 * torch.randn(100, 1)

model = nn.Linear(1, 1)
loss_fn = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.1)

for epoch in range(50):
    pred = model(X)
    loss = loss_fn(pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch % 10 == 0:
        print(f"Epoch {epoch}: Loss = {loss.item():.4f}")
````



---

### Learning Rate Sensitivity

| Learning Rate | Result             |
| ------------- | ------------------ |
| Too large     | Divergence         |
| Optimal       | Fast convergence   |
| Too small     | Very slow learning |

---

### Advanced Enhancements

| Technique | Purpose                     |
| --------- | --------------------------- |
| Momentum  | Accelerate convergence      |
| Nesterov  | Lookahead gradient          |
| AdaGrad   | Adaptive learning rates     |
| RMSProp   | Normalize gradients         |
| Adam      | Combines momentum + RMSProp |

---

### Practical Guidelines

| Rule               | Reason                        |
| ------------------ | ----------------------------- |
| Start with Adam    | Stable default                |
| Tune learning rate | Most important hyperparameter |
| Use mini-batches   | Computational efficiency      |
| Monitor gradients  | Detect vanishing/exploding    |

---

### Key Takeaway

> **All neural network training is fundamentally the problem of efficient gradient descent on extremely complex surfaces.**
