## Adagrad Optimizer

### Overview

**Adagrad (Adaptive Gradient Algorithm)** is an adaptive optimization method that assigns **individual learning rates to each parameter** based on the history of their gradients.
It is especially effective for **sparse data and high-dimensional feature spaces**, such as NLP and recommender systems.

---

### Core Intuition

Parameters that receive **large gradients frequently** should learn more slowly.
Parameters that receive **small or infrequent gradients** should learn faster.

Adagrad enforces this by **accumulating past squared gradients** and using them to scale future updates.

---

### Mathematical Formulation

Let:

* $g_t$ = gradient at step *t*
* $\theta_t$ = parameter at step *t*

Accumulate squared gradients:

$$
G_t = G_{t-1} + g_t^2
$$

Parameter update:

$$
\theta_{t+1} = \theta_t - \frac{\eta}{\sqrt{G_t + \epsilon}} \cdot g_t
$$

Each parameter maintains its **own learning rate**:

$$
\text{effectivelr} = \frac{\eta}{\sqrt{G_t + \epsilon}}
$$

---

### Training Workflow

| Stage      | Operation                               |
| ---------- | --------------------------------------- |
| Forward    | Compute loss                            |
| Backward   | Compute gradients                       |
| Accumulate | Update running sum of squared gradients |
| Scale      | Adjust learning rate per parameter      |
| Update     | Apply scaled parameter step             |

---

### Why Adagrad Works

| Property                    | Benefit                            |
| --------------------------- | ---------------------------------- |
| Per-parameter learning rate | Handles uneven feature frequencies |
| Automatic decay             | Stabilizes training                |
| Sparse features             | Strong performance in NLP, IR      |

---

### Limitations

| Issue              | Explanation                     |
| ------------------ | ------------------------------- |
| Aggressive decay   | Learning rate shrinks too fast  |
| Training stalls    | Long training becomes very slow |
| Poor for deep CNNs | Modern optimizers outperform    |

---

### PyTorch Demonstration

#### Dataset and Model



````python
import torch
import torch.nn as nn
import torch.optim as optim

model = nn.Sequential(
    nn.Linear(10, 50),
    nn.ReLU(),
    nn.Linear(50, 1)
)

criterion = nn.MSELoss()
````



#### Training with SGD



````python
opt_sgd = optim.SGD(model.parameters(), lr=0.1)
````



#### Training with Adagrad



````python

opt_adagrad = optim.Adagrad(model.parameters(), lr=0.1)
````



#### Training Loop



````python
for epoch in range(100):
    x = torch.randn(64, 10)
    y = torch.randn(64, 1)

    opt_adagrad.zero_grad()
    loss = criterion(model(x), y)
    loss.backward()
    opt_adagrad.step()
````



---

### Behavior Comparison

| Optimizer   | Learning Dynamics                |
| ----------- | -------------------------------- |
| SGD         | Constant learning rate           |
| Momentum    | Smoothed gradients               |
| **Adagrad** | Rapid early learning, slow later |

---

### Variants and Related Optimizers

| Optimizer | Key Improvement                    |
| --------- | ---------------------------------- |
| Adadelta  | Fixes Adagrad’s aggressive decay   |
| RMSProp   | Uses moving average instead of sum |
| Adam      | Combines momentum + RMSProp        |
| AdamW     | Proper weight decay handling       |

---

### When to Use Adagrad

| Scenario        | Suitability       |
| --------------- | ----------------- |
| Sparse features | Excellent         |
| NLP, embeddings | Very good         |
| Deep CNNs       | Not recommended   |
| Long training   | Often problematic |

---

### Key Takeaway

**Adagrad introduced the concept of adaptive per-parameter learning rates**, forming the foundation of modern optimizers such as RMSProp and Adam.
