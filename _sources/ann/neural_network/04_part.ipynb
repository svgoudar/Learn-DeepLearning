## Multi-Layer Perceptron (MLP) 

---

### **1. What is an MLP?**

A **Multi-Layer Perceptron (MLP)** is a **feed-forward neural network** composed of:

> **Input Layer → One or more Hidden Layers → Output Layer**

Each layer contains neurons connected by **learnable weights** and **biases**, followed by **non-linear activation functions**.

MLPs approximate arbitrary continuous functions and form the backbone of modern deep learning.

---

### **2. Mathematical Formulation**

For layer $$l$$:

$$
z^{(l)} = W^{(l)} a^{(l-1)} + b^{(l)}
$$
$$
a^{(l)} = \sigma(z^{(l)})
$$

Where:

| Symbol      | Meaning             |
| ----------- | ------------------- |
| $W^{(l)}$ | Weight matrix       |
| $b^{(l)}$ | Bias vector         |
| $\sigma$ | Activation function |
| $a^{(l)}$ | Output of layer     |

---

### **3. Why MLP Works**

| Property                | Explanation                                  |
| ----------------------- | -------------------------------------------- |
| Universal Approximation | Can approximate any continuous function      |
| Non-linearity           | Activation functions enable complex patterns |
| Hierarchical Learning   | Deeper layers learn higher-level features    |

---

### **4. Training Workflow**

1. **Forward Propagation**
2. **Loss Computation**
3. **Backpropagation (Chain Rule)**
4. **Weight Update (Optimizer)**

---

### **5. Activation Functions in MLP**

| Activation | Role                           |
| ---------- | ------------------------------ |
| ReLU       | Prevents vanishing gradients   |
| Sigmoid    | Binary classification output   |
| Tanh       | Centered nonlinearity          |
| Softmax    | Multi-class probability output |

---

### **6. Variants of MLP**

| Variant       | Description         |
| ------------- | ------------------- |
| Shallow MLP   | 1 hidden layer      |
| Deep MLP      | Many hidden layers  |
| Residual MLP  | Skip connections    |
| Dropout MLP   | Regularized network |
| BatchNorm MLP | Stabilized training |

---

### **7. PyTorch Demonstration**

#### **Model Definition**

```python
import torch
import torch.nn as nn

class MLP(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, hidden_dim),
            nn.ReLU(),
            nn.Linear(hidden_dim, output_dim))

    def forward(self, x):
        return self.net(x)
```

---

#### **Training Loop**

```python
model = MLP(10, 64, 1)
criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

for epoch in range(100):
    x = torch.randn(32, 10)
    y = torch.sum(x, dim=1, keepdim=True)

    pred = model(x)
    loss = criterion(pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
```

---

### **8. Interpretation of Learning**

| Layer        | Learns                         |
| ------------ | ------------------------------ |
| Early layers | Low-level feature combinations |
| Deep layers  | High-level abstractions        |

### **9. When to Use MLP**

| Task Type    | Suitability              |
| ------------ | ------------------------ |
| Tabular data | Excellent                |
| Time series  | With feature engineering |
| Vision       | As classifier head       |
| NLP          | As output projection     |

---

### **10. Limitations**

| Issue                   | Reason                   |
| ----------------------- | ------------------------ |
| Parameter explosion     | Fully connected          |
| No spatial awareness    | Unlike CNNs              |
| Sequential inefficiency | Unlike RNNs/Transformers |

---

**Summary**

> An MLP is the fundamental deep learning architecture that learns complex functions through stacked affine transformations and non-linear activations, trained end-to-end via backpropagation.
