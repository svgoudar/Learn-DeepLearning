```
[Area, Beds, Baths, Age, Location]
              ↓
          Linear Neurons
              ↓
         Nonlinear Hidden Layers
              ↓
           Price Output
```

---

### Learning Workflow

1. Forward pass: predict price
2. Compute error:
   $$
   \mathcal{L} = | \hat{y} - y |^2
   $$
3. Backpropagation: compute gradients
4. Update weights with gradient descent
5. Repeat until convergence

---

### PyTorch Implementation

#### Dataset (Synthetic)

```python
import torch
from torch import nn

X = torch.tensor([
    [1200, 3, 2, 10, 7],
    [800, 2, 1, 20, 5],
    [1500, 4, 3, 5, 9],
    [600, 1, 1, 30, 4]
], dtype=torch.float32)

y = torch.tensor([[300], [180], [420], [130]], dtype=torch.float32)
```

#### Model Definition

```python
class HouseNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(5, 16),
            nn.ReLU(),
            nn.Linear(16, 8),
            nn.ReLU(),
            nn.Linear(8, 1))

    def forward(self, x):
        return self.net(x)
```

#### Training Loop

```python
model = HouseNet()
loss_fn = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=0.01)

for epoch in range(1000):
    pred = model(X)
    loss = loss_fn(pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch % 200 == 0:
        print(f"Epoch {epoch}, Loss {loss.item():.4f}")
```

#### Inference

```python
new_house = torch.tensor([[1000, 3, 2, 8, 6]], dtype=torch.float32)
price = model(new_house)
print("Predicted price:", price.item())
```

### Why Nonlinearity Matters

Without activations, the model is only linear:
$$
f(x) = Wx + b
$$

With neurons + nonlinearities:

* Captures market interactions
* Models diminishing returns
* Learns feature coupling (e.g., large house in bad location ≠ expensive)
