---

### Convolutional Layer

**Intuition**

Learns spatial patterns using local receptive fields.
Shared weights drastically reduce parameters.

$$
y_{i,j} = \sum_{m,n} W_{m,n} \cdot x_{i+m,j+n}
$$

**Workflow**

1. Slide kernel over input
2. Compute dot products
3. Generate feature maps

**Remediation**

| Issue               | Fix                      |
| ------------------- | ------------------------ |
| Overfitting         | Data augmentation        |
| Poor generalization | BatchNorm, deeper stacks |

**PyTorch**

```python
conv = nn.Conv2d(in_channels=3, out_channels=32, kernel_size=3, padding=1)
```

---

### Recurrent Layers (RNN, LSTM, GRU)

**Intuition**

Maintain internal memory for sequential dependencies.

$$
h_t = f(x_t, h_{t-1})
$$

**Workflow**

1. Input at each timestep
2. Update hidden state
3. Propagate through time

**Remediation**

| Problem             | Fix               |
| ------------------- | ----------------- |
| Vanishing gradients | LSTM / GRU        |
| Exploding gradients | Gradient clipping |

**PyTorch**

```python
rnn = nn.LSTM(input_size=50, hidden_size=128, batch_first=True)
```

---

### Normalization Layers

**Intuition**

Reduce internal covariate shift.
Stabilize gradient flow.

| Type         | Usage          |
| ------------ | -------------- |
| BatchNorm    | CNNs           |
| LayerNorm    | Transformers   |
| InstanceNorm | Style transfer |

**PyTorch**

```python
bn = nn.BatchNorm2d(32)
ln = nn.LayerNorm(128)
```

---

### Activation Layers

**Intuition**

Introduce non-linearity enabling deep representation learning.

| Activation | Property                     |
| ---------- | ---------------------------- |
| ReLU       | Sparse, fast                 |
| GELU       | Smooth, Transformer standard |
| Sigmoid    | Probabilities                |
| Tanh       | Bounded output               |

**PyTorch**

```python
act = nn.ReLU()
```

---

### Pooling Layers

**Intuition**

Reduce spatial resolution while preserving semantics.

| Pool    | Effect                |
| ------- | --------------------- |
| MaxPool | Dominant features     |
| AvgPool | Smooth representation |

**PyTorch**

```python
pool = nn.MaxPool2d(2)
```

---

### Dropout Layer

**Intuition**

Randomly disables neurons to improve generalization.

$$
h' = h \odot \text{Bernoulli}(p)
$$

**PyTorch**

```python
drop = nn.Dropout(p=0.5)
```

---

### Attention Layer

**Intuition**

Allow each token to attend to all others.

$$
\text{Attention}(Q,K,V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d}}\right)V
$$

**PyTorch**

```python
attn = nn.MultiheadAttention(embed_dim=256, num_heads=8, batch_first=True)
```

---

### End-to-End Example Network

```python
class SimpleCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(),
            nn.MaxPool2d(2),
            nn.Flatten(),
            nn.Linear(32 * 16 * 16, 128),
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(128, 10)
        )

    def forward(self, x):
        return self.net(x)
```

---

### Layer Design Strategy

| Goal           | Recommended Layers           |
| -------------- | ---------------------------- |
| Vision         | Conv → Norm → ReLU → Pool    |
| NLP            | Embedding → Attention → Norm |
| Tabular        | Dense → ReLU → Dropout       |
| Stability      | Normalization everywhere     |
| Generalization | Dropout + Data augmentation  |

---

### Summary

Deep learning performance is fundamentally driven by **layer composition**.
Correct layer selection ensures:

* Efficient representation learning
* Stable optimization
* Robust generalization
