```
Input → Linear → Activation → ... → Linear → Output
```

**Applications**

Tabular data, regression, classification, feature modeling.

**Limitations**

| Issue                  | Cause         | Remediation                |
| ---------------------- | ------------- | -------------------------- |
| Vanishing gradients    | Deep chains   | ReLU, residual connections |
| Overfitting            | High capacity | Regularization, dropout    |
| Poor sequence modeling | No memory     | Use RNN/Transformer        |

**PyTorch**

```python
class MLP(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.net = torch.nn.Sequential(
            torch.nn.Linear(10, 64),
            torch.nn.ReLU(),
            torch.nn.Linear(64, 1)
        )
    def forward(self, x): return self.net(x)
```

### Convolutional Neural Networks (CNN)

**Intuition**

Convolutions detect **local spatial patterns** with shared weights.

**Workflow**

```
Image → Conv → ReLU → Pool → Conv → ... → FC → Output
```

**Applications**

Vision, medical imaging, remote sensing.

**Limitations & Remediation**

| Problem                    | Fix                                  |
| -------------------------- | ------------------------------------ |
| Loss of spatial resolution | Skip connections (ResNet, U-Net)     |
| Large parameter count      | Depthwise separable conv (MobileNet) |

**PyTorch**

```python
class CNN(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.conv = torch.nn.Conv2d(3, 16, 3)
        self.fc = torch.nn.Linear(16*30*30, 10)
    def forward(self, x):
        x = torch.relu(self.conv(x))
        return self.fc(x.view(x.size(0), -1))
```

### Recurrent Neural Networks (RNN, LSTM, GRU)

**Intuition**

RNNs maintain **temporal memory**.

$$
h_t = f(W_x x_t + W_h h_{t-1})
$$

**Workflow**

```
Sequence → RNN Cell → Hidden State → Next Step
```

**Limitations**

| Problem            | Remediation                   |
| ------------------ | ----------------------------- |
| Vanishing gradient | LSTM / GRU                    |
| Slow training      | Parallel models (Transformer) |

**PyTorch**

```python
rnn = torch.nn.LSTM(input_size=32, hidden_size=64, batch_first=True)
```

---

### Transformer Networks

**Intuition**

Self-attention learns **global dependencies** without recurrence.

$$
Attention(Q,K,V) = softmax(QK^T/\sqrt{d})V
$$

**Workflow**

```
Embedding → Self-Attention → FFN → Repeat → Output
```

**Applications**

NLP, vision, protein modeling, multimodal systems.

**Limitations & Fixes**

| Problem          | Fix                              |
| ---------------- | -------------------------------- |
| Quadratic memory | Sparse attention, FlashAttention |
| Context length   | Long-form transformers           |

**PyTorch**

```python
encoder = torch.nn.TransformerEncoderLayer(d_model=512, nhead=8)
```

### Autoencoders

**Intuition**

Learn compressed latent representations.

```
Input → Encoder → Latent → Decoder → Reconstruct
```

**Variants**

| Type              | Use                   |
| ----------------- | --------------------- |
| Denoising         | Noise-robust features |
| Variational (VAE) | Generative modeling   |

**PyTorch**

```python
class AutoEncoder(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.enc = torch.nn.Linear(100, 32)
        self.dec = torch.nn.Linear(32, 100)
```

---

### Graph Neural Networks (GNN)

**Intuition**

Learn on graph structure using message passing.

$$
h_v^{(k+1)} = \sigma(\sum_{u \in N(v)} W h_u^{(k)})
$$

**Applications**

Social networks, molecules, knowledge graphs.

---

### Generative Adversarial Networks (GAN)

**Intuition**

Two networks compete:

```
Generator vs Discriminator
```

**Variants**

DCGAN, CycleGAN, StyleGAN.

---

### Diffusion Models

**Intuition**

Learn to **reverse noise process**.

Used in: Stable Diffusion, DALL·E.

---

### Hybrid Architectures

| Architecture      | Purpose             |
| ----------------- | ------------------- |
| CNN + LSTM        | Video, speech       |
| CNN + Transformer | Vision transformers |
| GNN + Transformer | Graph reasoning     |

### Architectural Selection Guide

| Data Type   | Recommended Model   |
| ----------- | ------------------- |
| Tabular     | MLP                 |
| Image       | CNN, ViT            |
| Text        | Transformer         |
| Time series | LSTM, Transformer   |
| Graph       | GNN                 |
| Generation  | VAE, GAN, Diffusion |
