#### Model

```python
model = nn.Sequential(
    nn.Linear(2, 64),
    nn.ReLU(),
    nn.Linear(64, 2)
)
```

#### Training with SGD

```python
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=0.05)

for epoch in range(300):
    optimizer.zero_grad()
    outputs = model(X)
    loss = criterion(outputs, y)
    loss.backward()
    optimizer.step()
```

### SGD Variants

| Variant        | Key Idea                     |
| -------------- | ---------------------------- |
| Batch GD       | Uses entire dataset          |
| Stochastic GD  | One sample per update        |
| Mini-batch SGD | Practical standard           |
| SGD + Momentum | Accelerates convergence      |
| Nesterov       | Lookahead momentum           |
| AdaGrad        | Per-parameter learning rates |
| RMSProp        | Adaptive normalization       |
| Adam           | Momentum + adaptive rates    |

---

### SGD with Momentum (PyTorch)

```python
optimizer = optim.SGD(model.parameters(), lr=0.05, momentum=0.9)
```

**Update rule:**

$$
v_{t+1} = \beta v_t + \nabla_\theta \mathcal{L}
$$
$$
\theta_{t+1} = \theta_t - \eta v_{t+1}
$$

---

### Practical Hyperparameters

| Parameter     | Typical Range |
| ------------- | ------------- |
| Learning rate | 1e-4 – 1e-1   |
| Batch size    | 32 – 512      |
| Momentum      | 0.8 – 0.99    |
| Weight decay  | 1e-5 – 1e-3   |

---

### When to Prefer SGD over Adam

| Scenario                | Recommendation |
| ----------------------- | -------------- |
| Large datasets          | SGD            |
| Computer vision         | SGD            |
| Final training stage    | SGD            |
| Generalization critical | SGD            |

SGD often converges more slowly but yields **better generalization** than adaptive methods.

### Summary

| Aspect            | Insight                                        |
| ----------------- | ---------------------------------------------- |
| Purpose           | Minimize loss via noisy gradients              |
| Key benefit       | Scalability & generalization                   |
| Modern practice   | Mini-batch SGD with momentum                   |
| Foundational role | Basis of nearly all deep learning optimization |

### Final Intuition

> **SGD trades exactness for speed and generalization.**
> The noise in its updates is not a bug—it is the engine of deep learning.
