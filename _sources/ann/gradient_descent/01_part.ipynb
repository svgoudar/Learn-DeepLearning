## Batch Gradient Descent in Deep Learning

### Concept Overview

**Batch Gradient Descent (BGD)** is an optimization algorithm that updates model parameters using the **entire training dataset** to compute the gradient of the loss function at each step.

It is the most **theoretically stable** and **deterministic** form of gradient descent.

---

### Mathematical Formulation

For model parameters ( \theta ) and dataset ( D = {(x_i, y_i)}_{i=1}^N ):

[
\theta_{t+1} = \theta_t - \eta \nabla_\theta \left( \frac{1}{N} \sum_{i=1}^{N} \mathcal{L}(f(x_i; \theta), y_i) \right)
]

Where:

* ( \eta ) = learning rate
* ( \mathcal{L} ) = loss function
* ( f(x_i; \theta) ) = model

---

### Training Workflow

| Step             | Description                             |
| ---------------- | --------------------------------------- |
| Forward pass     | Compute predictions for **all samples** |
| Loss computation | Compute global loss                     |
| Backward pass    | Compute gradient over entire dataset    |
| Parameter update | Apply one update per epoch              |

---

### Key Characteristics

| Property         | Batch Gradient Descent  |
| ---------------- | ----------------------- |
| Stability        | Very high               |
| Convergence      | Smooth                  |
| Memory usage     | High                    |
| Speed per update | Slow                    |
| Scalability      | Poor for large datasets |
| Noise in updates | None                    |

---

### Comparison with Other Variants

| Method        | Data Used per Update | Stability | Speed               |
| ------------- | -------------------- | --------- | ------------------- |
| Batch GD      | Entire dataset       | High      | Slow                |
| Stochastic GD | 1 sample             | Noisy     | Fast                |
| Mini-batch GD | Small batch          | Balanced  | Optimal in practice |

---

### PyTorch Demonstration

#### Dataset & Model



````python
import torch
import torch.nn as nn
import torch.optim as optim

X = torch.randn(1000, 10)
y = torch.randn(1000, 1)

model = nn.Linear(10, 1)
loss_fn = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)
````


#### Training with Batch Gradient Descent



````python
for epoch in range(50):

    optimizer.zero_grad()

    predictions = model(X)          # All data
    loss = loss_fn(predictions, y)

    loss.backward()                 # Global gradient
    optimizer.step()                # One update per epoch

    print(f"Epoch {epoch} | Loss: {loss.item():.4f}")
````



This performs **one gradient update per epoch**, using the **entire dataset**.

---

### When to Use Batch Gradient Descent

| Use Case             | Reason                 |
| -------------------- | ---------------------- |
| Small datasets       | Fits in memory         |
| Convex optimization  | Guaranteed convergence |
| Theoretical analysis | Deterministic behavior |
| Debugging            | Stable gradients       |

---

### Limitations in Deep Learning

| Limitation               | Explanation                    |
| ------------------------ | ------------------------------ |
| Memory intensive         | Must load full dataset         |
| Slow updates             | One update per epoch           |
| Poor scalability         | Inefficient for large datasets |
| Local minima sensitivity | Can get stuck                  |

---

### Practical Insight

Modern deep learning rarely uses pure Batch Gradient Descent.
**Mini-batch Gradient Descent** dominates because it balances stability, speed, and scalability.

---

### Summary

| Aspect           | Insight                           |
| ---------------- | --------------------------------- |
| Update rule      | One update using all data         |
| Behavior         | Smooth, deterministic convergence |
| Best for         | Small-scale, convex problems      |
| Not suitable for | Large deep learning systems       |

---

### Final Intuition

> Batch Gradient Descent is the **foundation** of all gradient-based learning, but real-world deep learning relies on mini-batches for practical efficiency.
