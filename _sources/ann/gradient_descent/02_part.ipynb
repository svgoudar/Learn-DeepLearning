## Mini-Batch Gradient Descent 


**Mini-Batch Gradient Descent (MBGD)** is an optimization strategy where model parameters are updated using the gradient computed on a **small subset (mini-batch)** of the training data rather than the full dataset (Batch GD) or a single example (SGD).

It is the **default training paradigm** in modern deep learning because it balances:

* computational efficiency,
* gradient stability,
* and hardware utilization.

---

### Mathematical Formulation

Given loss function $\mathcal{L}(\theta)$ over dataset $D$,

For a mini-batch $B \subset D$:

$$
\theta_{t+1} = \theta_t - \eta \nabla_\theta \left$\frac{1}{|B|} \sum_{(x,y) \in B} \ell(f_\theta(x), y) \right)
$$

where:

* $\theta$ — parameters,
* $\eta$ — learning rate,
* $|B|$ — mini-batch size.

---

### Why Mini-Batch Works Best

| Method            | Stability    | Speed    | Hardware Utilization |
| ----------------- | ------------ | -------- | -------------------- |
| Batch GD          | High         | Slow     | Poor                 |
| SGD               | Noisy        | Fast     | Poor                 |
| **Mini-Batch GD** | **Balanced** | **Fast** | **Excellent**        |

---

### Training Workflow

1. Shuffle dataset
2. Split into mini-batches
3. For each batch:

   * Forward pass
   * Compute loss
   * Backpropagate gradients
   * Update parameters
4. Repeat for multiple epochs

---

### PyTorch Demonstration

#### Dataset & Model



````python
import torch
import torch.nn as nn
import torch.optim as optim
from sklearn.datasets import make_moons
from torch.utils.data import DataLoader, TensorDataset

X, y = make_moons(n_samples=2000, noise=0.15)
X = torch.tensor(X, dtype=torch.float32)
y = torch.tensor(y, dtype=torch.long)

loader = DataLoader(TensorDataset(X, y), batch_size=32, shuffle=True)

model = nn.Sequential(
    nn.Linear(2, 64),
    nn.ReLU(),
    nn.Linear(64, 2)
)

loss_fn = nn.CrossEntropyLoss()
optimizer = optim.Adam(model.parameters(), lr=0.01)
````



---

#### Training Loop (Mini-Batch GD)



````python
for epoch in range(50):
    for xb, yb in loader:
        preds = model(xb)
        loss = loss_fn(preds, yb)

        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
````



---

### Intuition Behind Mini-Batches

| Component                | Role                               |
| ------------------------ | ---------------------------------- |
| Noise in gradients       | Helps escape sharp local minima    |
| Smaller memory footprint | Enables training on large datasets |
| Parallel computation     | Exploits GPU efficiently           |
| More frequent updates    | Faster convergence                 |

---

### Effect of Batch Size

| Batch Size      | Behavior                               |
| --------------- | -------------------------------------- |
| Small (16–32)   | Noisy but generalizes well             |
| Medium (64–256) | Balanced                               |
| Large (1024+)   | Stable but risk of poor generalization |

---

### Variants & Enhancements

| Technique                | Purpose                 |
| ------------------------ | ----------------------- |
| Momentum                 | Smooths oscillations    |
| Nesterov                 | Look-ahead updates      |
| Adam / RMSProp           | Adaptive learning rates |
| Gradient Accumulation    | Simulates large batches |
| Learning Rate Scheduling | Faster convergence      |

---

### Common Pitfalls

* Too large batch → poor generalization
* Too small batch → unstable training
* Inadequate shuffling → biased updates

---

### Summary

| Property          | Mini-Batch GD |
| ----------------- | ------------- |
| Update frequency  | High          |
| Memory usage      | Moderate      |
| Convergence speed | Fast          |
| Scalability       | Excellent     |

> **Mini-Batch Gradient Descent is the practical optimization backbone of modern deep learning systems.**
