## Gradient Descent

### Concept and Intuition

**Gradient Descent (GD)** is the core optimization algorithm used to train deep neural networks.
Its purpose is to **minimize a loss function** by iteratively adjusting model parameters in the direction of **steepest decrease**.

Intuitively, GD treats the loss surface as a high-dimensional landscape:

* Height → loss value
* Position → model parameters
* Slope → gradient

At each step, GD moves the parameters **downhill**.

---

### Mathematical Formulation

Given a model with parameters $\theta$ and loss function $\mathcal{L}(\theta)$:

$$
\theta_{t+1} = \theta_t - \eta \nabla_{\theta} \mathcal{L}(\theta_t)
$$

Where:

| Symbol                          | Meaning              |
| ------------------------------- | -------------------- |
| $\eta$                        | Learning rate        |
| $\nabla_{\theta} \mathcal{L}$ | Gradient of the loss |
| $\theta_t$                    | Current parameters   |

---

### Training Workflow

| Step             | Operation              |
| ---------------- | ---------------------- |
| Forward pass     | Compute predictions    |
| Loss computation | Measure error          |
| Backward pass    | Compute gradients      |
| Update step      | Apply gradient descent |
| Repeat           | Until convergence      |

---

### Variants of Gradient Descent

| Variant       | Data Used per Step | Characteristics   |
| ------------- | ------------------ | ----------------- |
| Batch GD      | Entire dataset     | Stable but slow   |
| Stochastic GD | One sample         | Fast, noisy       |
| Mini-batch GD | Small batch        | Industry standard |

---

### PyTorch Demonstration

#### Example: Linear Regression with Gradient Descent



````python
import torch
import torch.nn as nn
import torch.optim as optim

# Data
x = torch.randn(100, 1)
y = 3 * x + 2 + 0.1 * torch.randn(100, 1)

# Model
model = nn.Linear(1, 1)

# Loss and Optimizer
loss_fn = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.1)

# Training Loop
for epoch in range(50):
    y_pred = model(x)
    loss = loss_fn(y_pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if epoch % 10 == 0:
        print(f"Epoch {epoch}, Loss: {loss.item():.4f}")
````



---

### How PyTorch Implements Gradient Descent

| Operation            | PyTorch Mechanism |
| -------------------- | ----------------- |
| Gradient computation | Autograd          |
| Parameter update     | Optimizer.step()  |
| Graph clearing       | zero_grad()       |

### Advanced Optimization Extensions

| Method   | Key Idea                            |
| -------- | ----------------------------------- |
| Momentum | Accelerates in consistent direction |
| RMSProp  | Normalizes gradient magnitude       |
| Adam     | Momentum + adaptive scaling         |
| Adagrad  | Learning rate per parameter         |

---

### Learning Rate Effects

| Learning Rate | Behavior                |
| ------------- | ----------------------- |
| Too small     | Slow convergence        |
| Too large     | Divergence              |
| Well tuned    | Fast stable convergence |

---

### Convergence Characteristics

GD converges to:

* **Global minimum** for convex problems
* **Local minimum / saddle point** for deep networks

Deep learning success relies on navigating this non-convex landscape effectively.

---

### Summary

| Concept         | Role                         |
| --------------- | ---------------------------- |
| Gradient        | Direction of steepest ascent |
| Descent         | Move opposite to gradient    |
| Optimizer       | Controls step strategy       |
| Learning rate   | Controls step size           |
| Backpropagation | Computes gradients           |

---

### Final Insight

> **Training a neural network is fundamentally the process of performing gradient descent in a very high-dimensional space.**
