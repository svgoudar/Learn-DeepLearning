
## ReLU and GELU Activation Functions

### Role of Activation Functions

Neural networks are compositions of linear transformations. Without **activation functions**, stacked layers collapse into a single linear model. Activations introduce **non-linearity**, allowing networks to learn complex functions.

$$
\mathbf{y} = \sigma(\mathbf{W}\mathbf{x} + \mathbf{b})
$$

where $\sigma$ is the activation.

---

### ReLU — Rectified Linear Unit

**Definition**

$$
\text{ReLU}(x) = \max(0, x)
$$

**Intuition**

* Acts as a **gate**: only positive signals propagate.
* Encourages **sparse representations** — many neurons output zero.
* Provides strong gradients for positive activations → fast convergence.

**Gradient**

$$
\frac{d}{dx}\text{ReLU}(x) =
\begin{cases}
1, & x > 0 \
0, & x \le 0
\end{cases}
$$

**Advantages**

| Property              | Effect                     |
| --------------------- | -------------------------- |
| Fast convergence      | Avoids vanishing gradients |
| Sparse activations    | Efficient representation   |
| Computationally cheap | Simple max operation       |

**Failure Mode: Dying ReLU**

When neurons enter the negative region permanently:

$$
x \le 0 \Rightarrow \text{ReLU}(x)=0,\quad \nabla=0
$$

Neuron stops learning.

**Remediation**

* Proper initialization (He initialization)
* Smaller learning rate
* Leaky ReLU / ELU variants

---

### GELU — Gaussian Error Linear Unit

**Definition**

$$
\text{GELU}(x) = x \cdot \Phi(x)
$$

where $\Phi(x)$ is the CDF of the standard normal distribution.

Approximation (used in practice):

$$
\text{GELU}(x) \approx 0.5x\left(1+\tanh\left(\sqrt{\frac{2}{\pi}}(x+0.044715x^3)\right)\right)
$$

**Intuition**

Instead of hard-gating like ReLU, GELU performs **soft probabilistic gating**:

* Inputs are weighted by their likelihood of being useful.
* Negative inputs are not zeroed — they are smoothly attenuated.

**Properties**

| Property                | Effect                |
| ----------------------- | --------------------- |
| Smooth & differentiable | Stable optimization   |
| Non-zero for negatives  | Prevents dead neurons |
| Probabilistic gating    | Better expressiveness |

**Why Transformers Use GELU**

* Maintains gradient flow for all neurons
* Improves representation quality in deep attention stacks
* Empirically superior for language models

---

### Comparison Summary

| Feature              | ReLU     | GELU     |
| -------------------- | -------- | -------- |
| Smooth               | No       | Yes      |
| Zero for negatives   | Yes      | No       |
| Dying neuron risk    | High     | None     |
| Compute cost         | Very low | Moderate |
| Used in CNNs         | ✓        | Rare     |
| Used in Transformers | Rare     | ✓        |

---

### Workflow in Neural Training

1. Linear transformation
2. Activation (ReLU or GELU)
3. Loss computation
4. Backpropagation through activation
5. Weight updates

The activation directly controls gradient propagation and therefore **learning dynamics**.

---

### PyTorch Demonstration

```python
import torch
import torch.nn as nn

x = torch.linspace(-3, 3, 100)

relu = nn.ReLU()
gelu = nn.GELU()

y_relu = relu(x)
y_gelu = gelu(x)
```

Neural network usage:

```python
class MLP(nn.Module):
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(
            nn.Linear(128, 256),
            nn.GELU(),
            nn.Linear(256, 10)
        )

    def forward(self, x):
        return self.net(x)
```

### Variants and Extensions

| Variant    | Purpose                  |
| ---------- | ------------------------ |
| Leaky ReLU | Fix dying ReLU           |
| PReLU      | Learnable negative slope |
| ELU        | Faster convergence       |
| Swish      | Similar to GELU          |
| Mish       | Smooth non-monotonic     |

---

### Practical Guidelines

| Scenario                | Recommended       |
| ----------------------- | ----------------- |
| CNNs / vision           | ReLU / Leaky ReLU |
| Transformers / NLP      | GELU              |
| Very deep networks      | GELU              |
| Low-resource / embedded | ReLU              |

---

### Key Takeaways

* **ReLU** = hard gate → fast, simple, but fragile.
* **GELU** = probabilistic soft gate → stable, expressive, modern.
* Activation choice directly impacts **trainability, stability, and final performance**.
