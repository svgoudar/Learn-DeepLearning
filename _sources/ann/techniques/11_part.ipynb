
## Momentum-Based Training

**Momentum** is an optimization technique designed to accelerate gradient descent by accumulating a velocity vector in parameter space.
Instead of updating parameters solely based on the current gradient, momentum integrates information from **past gradients**, enabling:

* Faster convergence
* Reduced oscillations in narrow valleys
* Better navigation of ravines and plateaus

**Physical Analogy**

Imagine a ball rolling down a landscape:

* Gradient descent: the ball reacts only to the current slope.
* Momentum: the ball accumulates speed, helping it roll through small bumps and shallow minima.

---

### Mathematical Formulation

Let:

* $\theta$ be parameters
* $g_t = \nabla_\theta L(\theta_t)$ be the gradient
* $v_t$ be the velocity
* $\mu \in [0, 1)$ be the momentum coefficient
* $\eta$ be the learning rate

**Update rules:**

$$
v_t = \mu v_{t-1} + \eta g_t
$$
$$
\theta_{t+1} = \theta_t - v_t
$$

---

### Why Momentum Works

| Problem in Vanilla SGD | How Momentum Helps                         |
| ---------------------- | ------------------------------------------ |
| Slow convergence       | Accumulates consistent gradients           |
| Oscillation in ravines | Damps oscillations across steep directions |
| Local noise            | Averages gradient noise over time          |
| Plateaus               | Maintains velocity to escape flat regions  |

---

### Training Workflow with Momentum

1. **Initialize**

   * Parameters $\theta$
   * Velocity $v = 0$

2. **For each batch**

   * Compute gradient $g$
   * Update velocity $v = \mu v + \eta g$
   * Update parameters $\theta = \theta - v$

3. **Repeat until convergence**

---

### PyTorch Demonstration

```python
import torch
from torch import nn
from torch.optim import SGD

# Simple model
model = nn.Linear(10, 1)

# Momentum optimizer
optimizer = SGD(model.parameters(), lr=0.01, momentum=0.9)

loss_fn = nn.MSELoss()

for epoch in range(100):
    x = torch.randn(32, 10)
    y = torch.randn(32, 1)

    pred = model(x)
    loss = loss_fn(pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()
```

---

### Key Hyperparameters

| Parameter              | Typical Range | Effect                   |
| ---------------------- | ------------- | ------------------------ |
| Learning rate $\eta$ | 1e-4 to 1e-1  | Step size                |
| Momentum $\mu$       | 0.8 to 0.99   | Memory of past gradients |

---

### Variants of Momentum

| Variant            | Idea                                 | Update                           |
| ------------------ | ------------------------------------ | -------------------------------- |
| Classical Momentum | Accumulate past gradients            | $v_t = \mu v_{t-1} + \eta g_t$ |
| Nesterov Momentum  | Look ahead before computing gradient | $g(\theta - \mu v)$            |
| Adam               | Momentum + adaptive scaling          | Combines first & second moments  |
| RMSProp + Momentum | Scale + smooth                       | Adaptive step with velocity      |

---

### Common Failure Modes and Remediation

| Issue               | Cause                         | Fix                         |
| ------------------- | ----------------------------- | --------------------------- |
| Overshooting minima | High learning rate + momentum | Reduce learning rate        |
| Instability         | Momentum too high             | Lower $\mu$               |
| Slow start          | Small initial gradients       | Warmup learning rate        |
| Divergence          | Poor scaling                  | Normalize inputs / use Adam |

---

### When to Use Momentum

| Scenario                      | Benefit                 |
| ----------------------------- | ----------------------- |
| Deep neural networks          | Faster convergence      |
| Ill-conditioned loss surfaces | Stabilized optimization |
| Noisy gradients               | Reduced variance        |
| Large batch training          | Improved generalization |

---

### Summary

Momentum transforms gradient descent from a purely reactive process into a **dynamic system with memory**.
This allows neural networks to train **faster, smoother, and more reliably**, making momentum a foundational technique in modern deep learning optimization.
