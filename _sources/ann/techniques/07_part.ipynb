## Remediating Overfitting and Underfitting


In supervised learning, the model’s goal is to **generalize** — perform well on unseen data.

| Problem      | What is Happening                                | Mental Model                |
| ------------ | ------------------------------------------------ | --------------------------- |
| Underfitting | Model is too simple to capture structure         | *High bias*                 |
| Overfitting  | Model memorizes training data, fails on new data | *High variance*             |
| Good fit     | Balanced complexity and generalization           | *Bias–variance equilibrium* |

Learning curves reveal the problem:

| Pattern      | Training Error | Validation Error |
| ------------ | -------------- | ---------------- |
| Underfitting | High           | High             |
| Overfitting  | Low            | High             |
| Good Fit     | Low            | Low              |

---

### Workflow for Diagnosing Fit

1. Train model
2. Track training & validation loss/accuracy
3. Inspect learning curves
4. Classify regime: underfit, overfit, or good fit
5. Apply targeted remediation

---

### Underfitting: Causes and Remedies

#### Causes

* Model capacity too low
* Features too weak
* Excessive regularization
* Insufficient training time

#### Remediation Strategies

| Strategy                | Intuition                        |
| ----------------------- | -------------------------------- |
| Increase model capacity | Allow more expressive function   |
| Add layers / neurons    | Learn more complex patterns      |
| Reduce regularization   | Stop constraining learning       |
| Train longer            | Let optimization converge        |
| Improve features        | Give better input representation |

#### PyTorch Demonstration: Fixing Underfitting

```python
class SimpleNet(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.net = torch.nn.Sequential(
            torch.nn.Linear(10, 16),
            torch.nn.ReLU(),
            torch.nn.Linear(16, 1)
        )
```

Upgrade capacity:

```python
class BetterNet(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.net = torch.nn.Sequential(
            torch.nn.Linear(10, 64),
            torch.nn.ReLU(),
            torch.nn.Linear(64, 64),
            torch.nn.ReLU(),
            torch.nn.Linear(64, 1)
        )
```

### Overfitting: Causes and Remedies

#### Causes

* Model too complex for dataset size
* Too few training samples
* Noisy labels
* Weak regularization

#### Remediation Strategies

| Strategy          | Intuition                       |
| ----------------- | ------------------------------- |
| Regularization    | Penalize memorization           |
| Dropout           | Force redundant representations |
| Data augmentation | Increase effective dataset      |
| Early stopping    | Stop before memorization        |
| Simplify model    | Reduce capacity                 |
| Weight decay      | Constrain parameter growth      |

#### PyTorch Demonstration: Fixing Overfitting

```python
class OverfitNet(torch.nn.Module):
    def __init__(self):
        super().__init__()
        self.net = torch.nn.Sequential(
            torch.nn.Linear(10, 256),
            torch.nn.ReLU(),
            torch.nn.Linear(256, 256),
            torch.nn.ReLU(),
            torch.nn.Linear(256, 1)
        )
```

### Summary: Decision Table

| Symptom             | Apply               |
| ------------------- | ------------------- |
| High train error    | Increase capacity   |
| High val error only | Add regularization  |
| Both high           | Improve data, model |
| Large gap           | Reduce variance     |

---

### Conceptual Takeaway

> **Underfitting = insufficient learning capacity**
> **Overfitting = excess memorization**
> **Remediation = control model expressiveness relative to data**

This balance defines effective deep learning systems.
