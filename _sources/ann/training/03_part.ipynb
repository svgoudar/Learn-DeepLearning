
## Loss Function

### Definition and Role

A **loss function** $\mathcal{L}$ quantifies how far a model’s prediction $\hat{y}$ is from the true target $y$.
Training a neural network is the process of **minimizing the expected loss** over the data distribution.

$$
\min_\theta ; \mathbb{E}*{(x,y)\sim D}[\mathcal{L}(f*\theta(x), y)]
$$

where
$\theta$ = model parameters,
$f_\theta$ = neural network.

Loss functions convert learning into an **optimization problem**.

---

### Why Loss Functions Matter

The loss function determines:

* What the model **learns**
* How gradients **flow**
* Training **stability**
* Convergence **speed**
* Final **model behavior**

Poor loss choice ⇒ slow learning, unstable gradients, wrong objective.

---

### Core Components

| Component     | Meaning                                                |
| ------------- | ------------------------------------------------------ |
| Model output  | $\hat{y} = f_\theta(x)$                              |
| Ground truth  | $y$                                                  |
| Loss function | $\mathcal{L}(\hat{y}, y)$                            |
| Optimizer     | Updates $\theta$ using $\nabla_\theta \mathcal{L}$ |

---

### Main Classes of Loss Functions

#### Regression Losses

| Loss                      | Formula            | Properties                    |   |                    |
| ------------------------- | ------------------ | ----------------------------- | - | ------------------ |
| Mean Squared Error (MSE)  | $(y-\hat{y})^2$  | Smooth, sensitive to outliers |   |                    |
| Mean Absolute Error (MAE) | $                 | y-\hat{y}                     |$ | Robust to outliers |
| Huber                     | Quadratic + Linear | Stable & robust               |   |                    |

#### Classification Losses

| Loss                      | Used For              | Key Idea            |
| ------------------------- | --------------------- | ------------------- |
| Binary Cross-Entropy      | Binary classification | Log-likelihood      |
| Categorical Cross-Entropy | Multi-class           | Softmax likelihood  |
| Hinge                     | SVM-style             | Margin maximization |

#### Probabilistic Interpretation

Cross-entropy loss = **negative log-likelihood** of the correct class:

$$
\mathcal{L} = -\log P(y \mid x; \theta)
$$

Thus training performs **maximum likelihood estimation**.

---

### Geometry of Loss Landscape

Loss defines a high-dimensional surface over parameter space.

Training = navigating this surface using gradients:

$$
\theta_{t+1} = \theta_t - \eta \nabla_\theta \mathcal{L}
$$

Good loss → smooth landscape → stable gradients.

---

### PyTorch Demonstration

#### Regression Example — MSE Loss

```python
import torch
import torch.nn as nn

# Model
model = nn.Linear(1, 1)

# Data
x = torch.tensor([[1.0], [2.0], [3.0]])
y = torch.tensor([[2.0], [4.0], [6.0]])

# Loss & optimizer
criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# Training step
for _ in range(100):
    y_pred = model(x)
    loss = criterion(y_pred, y)

    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

print("Final loss:", loss.item())
```

---

#### Classification Example — Cross Entropy

```python
import torch
import torch.nn as nn

model = nn.Linear(10, 3)   # 3 classes

x = torch.randn(5, 10)
y = torch.tensor([0, 2, 1, 2, 0])  # class labels

criterion = nn.CrossEntropyLoss()

logits = model(x)
loss = criterion(logits, y)

loss.backward()
print("Loss:", loss.item())
```

`CrossEntropyLoss` internally applies **Softmax + Log + NLLLoss**.

### Custom Loss Example

```python
class MyLoss(nn.Module):
    def forward(self, y_pred, y):
        return torch.mean((y_pred - y)**4)

criterion = MyLoss()
```

### Practical Workflow

* Choose loss based on **task**
* Monitor loss curve for **convergence**
* Adjust loss if:

  * gradients vanish/explode
  * training unstable
  * evaluation metric mismatched

---

### Key Insight

> **The loss function defines the learning objective.
> The optimizer only follows the gradients the loss creates.**

The loss is the true "teacher" of the neural network.
