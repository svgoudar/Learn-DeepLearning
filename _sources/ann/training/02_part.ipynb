
## Model Definition

### Conceptual Meaning

A **model** in deep learning is a **parameterized mathematical function**
$$
f_\theta : \mathbb{R}^n \rightarrow \mathbb{R}^m
$$
that maps input data to outputs by composing multiple nonlinear transformations, where:

* $\theta$ = set of **learnable parameters** (weights & biases)
* The structure of $f$ = **architecture**
* Learning = optimizing $\theta$ from data

In practice, a model is implemented as a **computational graph** of tensor operations.

---

### Why Model Definition Matters

Model definition determines:

| Aspect               | Controlled By Model           |
| -------------------- | ----------------------------- |
| Representation power | Depth, width, activations     |
| Inductive bias       | CNN, RNN, Transformer, etc.   |
| Generalization       | Architecture + regularization |
| Computational cost   | Layer types & sizes           |

---

### Core Components of a Deep Learning Model

**1. Layers**
Each layer performs:
$$
h = \phi(Wx + b)
$$

* Linear transformation
* Nonlinear activation

**2. Parameters**

| Type    | Purpose                    |
| ------- | -------------------------- |
| Weights | Learn feature interactions |
| Biases  | Shift activation           |

**3. Forward Function**

Defines how data flows through the network.

**4. Loss Function**

Measures model error:
$$
\mathcal{L}(y, \hat{y})
$$

**5. Optimizer**

Updates parameters using gradients:
$$
\theta \leftarrow \theta - \eta \nabla_\theta \mathcal{L}
$$

---

### Mathematical View

$$
\hat{y} = f_\theta(x) = f_L(f_{L-1}(...f_1(x)))
$$

Each $f_i$ is a layer transformation.

---

### PyTorch Model Definition Workflow

| Step                    | Purpose                |
| ----------------------- | ---------------------- |
| Define class            | Architecture blueprint |
| `__init__()`            | Create layers          |
| `forward()`             | Define computation     |
| Instantiate model       | Create network         |
| Attach loss & optimizer | Enable training        |

---

### PyTorch Demonstration

#### Define Model

```python
import torch
import torch.nn as nn

class SimpleNN(nn.Module):
    def __init__(self, input_dim, hidden_dim, output_dim):
        super().__init__()
        self.fc1 = nn.Linear(input_dim, hidden_dim)
        self.relu = nn.ReLU()
        self.fc2 = nn.Linear(hidden_dim, output_dim)

    def forward(self, x):
        x = self.fc1(x)
        x = self.relu(x)
        x = self.fc2(x)
        return x
```

#### Instantiate Model

```python
model = SimpleNN(input_dim=10, hidden_dim=32, output_dim=1)
```

#### Inspect Parameters

```python
for name, param in model.named_parameters():
    print(name, param.shape)
```

---

### Data Flow Through the Model

```python
x = torch.randn(4, 10)   # batch of 4 samples
y_pred = model(x)
```

### Conceptual Summary

| Layer            | Role                      |
| ---------------- | ------------------------- |
| Architecture     | Defines hypothesis space  |
| Parameters       | Encodes learned knowledge |
| Forward function | Computes predictions      |
| Backward pass    | Computes gradients        |
| Optimizer        | Improves parameters       |

### Key Insight

> **Model definition specifies the space of functions the learning algorithm can explore.**
> Training only finds the best function within that space.
