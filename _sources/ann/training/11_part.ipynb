### Online / Continual Training

Model updated as data arrives.

```python
for X, y in data_stream:
    loss = criterion(model(X), y)
    loss.backward()
    optimizer.step()
```

Train on easier data first.

### Mixed Precision Training

Reduces memory and increases speed.

```python
scaler = torch.cuda.amp.GradScaler()

with torch.cuda.amp.autocast():
    loss = criterion(model(X), y)

scaler.scale(loss).backward()
scaler.step(optimizer)
scaler.update()
```

---

### Training Strategy Selection

| Scenario           | Strategy                 |
| ------------------ | ------------------------ |
| Small dataset      | Transfer learning        |
| Massive dataset    | Distributed training     |
| Limited GPU memory | Mixed precision          |
| Streaming data     | Online training          |
| High overfitting   | Curriculum + fine-tuning |

---

### Key Insight

> Training is not a single algorithm — it is a **system design problem** combining optimization, data management, and generalization control.
