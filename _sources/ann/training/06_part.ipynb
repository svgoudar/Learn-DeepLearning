## Forward Propagation 

---

### Conceptual Overview

**Forward propagation** is the process by which input data flows through a neural network to produce an output (prediction).
It is the **deterministic computation graph execution** defined by the model architecture and parameters.

Mathematically, for a feed-forward neural network:

$$
\mathbf{a}^{(0)} = \mathbf{x}
$$
$$
\mathbf{z}^{(l)} = \mathbf{W}^{(l)} \mathbf{a}^{(l-1)} + \mathbf{b}^{(l)}
$$
$$
\mathbf{a}^{(l)} = \phi^{(l)}(\mathbf{z}^{(l)})
$$

Where:

| Symbol                                 | Meaning                          |
| -------------------------------------- | -------------------------------- |
| $\mathbf{x}$                         | Input vector                     |
| $\mathbf{W}^{(l)}, \mathbf{b}^{(l)}$ | Weights & bias of layer (l)      |
| $\mathbf{z}^{(l)}$                   | Pre-activation                   |
| $\phi^{(l)}$                         | Activation function              |
| $\mathbf{a}^{(l)}$                   | Output (activation) of layer (l) |

---

### Intuition

Forward propagation is:

* **Function composition**
* **Feature transformation**
* **Information flow**
* **Prediction generation**

Each layer transforms its input representation into a more abstract representation.

---

### Computational Workflow

**Pipeline**

1. Receive input vector $\mathbf{x}$
2. For each layer:

   * Linear transform: $\mathbf{z} = \mathbf{W}\mathbf{a} + \mathbf{b}$
   * Non-linear activation: $\mathbf{a} = \phi(\mathbf{z})$
3. Final output $\hat{y}$ is produced

**Graph View**

```
Input → Linear → Activation → Linear → Activation → ... → Output
```

---

### Why Forward Propagation Matters

| Role                 | Purpose                                        |
| -------------------- | ---------------------------------------------- |
| Prediction           | Produces model output                          |
| Loss computation     | Supplies output for loss evaluation            |
| Gradient computation | Stores intermediate values for backpropagation |
| Learning             | Defines what the model currently "believes"    |

### Forward Propagation in PyTorch (Manual Illustration)



````python
import torch
import torch.nn.functional as F

# Input
x = torch.tensor([[1.0, 2.0]])

# Parameters
W1 = torch.randn(2, 3)
b1 = torch.randn(3)

W2 = torch.randn(3, 1)
b2 = torch.randn(1)

# Layer 1
z1 = x @ W1 + b1
a1 = F.relu(z1)

# Layer 2
z2 = a1 @ W2 + b2
y_hat = z2  # regression output

print(y_hat)
````



---

### Forward Propagation Using PyTorch Modules



````python
import torch
import torch.nn as nn

class SimpleNet(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(2, 3)
        self.fc2 = nn.Linear(3, 1)

    def forward(self, x):
        x = torch.relu(self.fc1(x))
        x = self.fc2(x)
        return x

model = SimpleNet()
x = torch.tensor([[1.0, 2.0]])
output = model(x)

print(output)
````



The `forward()` function explicitly defines the forward propagation logic.

### Common Activation Functions in Forward Pass

| Activation | Formula                            | Use Case                  |
| ---------- | ---------------------------------- | ------------------------- |
| ReLU       | $\max(0,z)$                      | Default for hidden layers |
| Sigmoid    | $\frac{1}{1+e^{-z}}$             | Binary output             |
| Tanh       | $\tanh(z)$                       | Centered activations      |
| Softmax    | $\frac{e^{z_i}}{\sum_j e^{z_j}}$ | Multiclass output         |

---

### Variants of Forward Propagation

**Fully Connected Networks**
Linear → Activation → Linear → Activation → …

**Convolutional Networks**
Conv → Activation → Pool → Conv → …

**Residual Networks**
$y = F(x) + x$

**Recurrent Networks**
Hidden state propagation over time:

$$
h_t = \phi(W_h h_{t-1} + W_x x_t)
$$

**Transformers**
Self-attention → Feed-forward → Layer norm

---

### Computational Properties

| Aspect         | Description                             |
| -------------- | --------------------------------------- |
| Deterministic  | Same input + same weights → same output |
| Differentiable | Enables gradient computation            |
| Vectorized     | GPU-efficient                           |
| Composable     | Forms deep computation graphs           |

---

### Relationship to Backpropagation

Forward propagation:

* Builds computation graph
* Stores intermediate activations

Backpropagation:

* Uses stored values to compute gradients
* Updates weights

---

### Summary

Forward propagation is the **core inference mechanism** of neural networks:

* Defines how inputs become predictions
* Establishes the structure for learning
* Enables gradient-based optimization

It is the computational foundation upon which **training**, **generalization**, and **deployment** are built.
