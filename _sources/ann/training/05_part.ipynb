## Training Loop 

---

### Definition

A **training loop** is the core algorithmic procedure that iteratively updates model parameters to minimize a loss function over a dataset using optimization methods (e.g., Gradient Descent).

Formally, it solves:

$$
\theta^* = \arg\min_\theta ; \mathbb{E}*{(x,y)\sim D} ; \mathcal{L}(f*\theta(x), y)
$$

---

### Conceptual Components

| Component     | Purpose                                |
| ------------- | -------------------------------------- |
| Model         | Parametric function $f_\theta$       |
| Dataset       | Provides samples $(x, y)$            |
| Loss Function | Measures prediction error              |
| Optimizer     | Updates parameters using gradients     |
| Forward Pass  | Computes predictions and loss          |
| Backward Pass | Computes gradients via backpropagation |
| Update Step   | Adjusts parameters                     |
| Epoch         | One full pass over dataset             |
| Batch         | Subset of data for one update          |

---

### Canonical Training Loop Workflow

1. Sample a batch from data
2. Forward pass: compute predictions
3. Compute loss
4. Backward pass: compute gradients
5. Update parameters
6. Clear gradients
7. Repeat until convergence

---

### Mathematical View

For each batch $B$:

$$
\hat{y} = f_\theta(x)
$$

$$
\mathcal{L} = \ell(\hat{y}, y)
$$

$$
\nabla_\theta \mathcal{L} = \frac{\partial \mathcal{L}}{\partial \theta}
$$

$$
\theta \leftarrow \theta - \eta \nabla_\theta \mathcal{L}
$$

---

### PyTorch Demonstration

```python
import torch
import torch.nn as nn
import torch.optim as optim

# Dataset
X = torch.randn(1000, 10)
y = torch.sum(X, dim=1, keepdim=True)

# Model
model = nn.Linear(10, 1)

# Loss and optimizer
criterion = nn.MSELoss()
optimizer = optim.SGD(model.parameters(), lr=0.01)

# Training loop
for epoch in range(20):
    optimizer.zero_grad()            # Clear old gradients
    y_pred = model(X)                # Forward pass
    loss = criterion(y_pred, y)      # Compute loss
    loss.backward()                  # Backward pass
    optimizer.step()                 # Parameter update

    print(f"Epoch {epoch} | Loss: {loss.item():.4f}")
```

---

### Full Training Loop with Batching and Evaluation

```python
from torch.utils.data import DataLoader, TensorDataset

dataset = TensorDataset(X, y)
loader = DataLoader(dataset, batch_size=32, shuffle=True)

for epoch in range(10):
    model.train()
    total_loss = 0

    for xb, yb in loader:
        optimizer.zero_grad()
        pred = model(xb)
        loss = criterion(pred, yb)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    model.eval()
    print(f"Epoch {epoch} | Avg Loss: {total_loss/len(loader):.4f}")
```

### Key Best Practices

| Practice                 | Reason                               |
| ------------------------ | ------------------------------------ |
| Zero gradients           | Prevent accumulation                 |
| Mini-batches             | Stable gradients, faster convergence |
| Shuffling                | Reduces overfitting                  |
| Evaluation mode          | Disables dropout, batchnorm update   |
| Learning rate scheduling | Controls convergence speed           |
| Gradient clipping        | Prevents exploding gradients         |

---

### Common Variants

| Variant              | Description                     |
| -------------------- | ------------------------------- |
| Online Training      | Batch size = 1                  |
| Mini-batch Training  | Most common approach            |
| Full Batch           | Uses entire dataset             |
| Distributed Training | Parallel across GPUs/nodes      |
| Mixed Precision      | FP16 for faster compute         |
| Curriculum Learning  | Train from easy to hard samples |

---

### Convergence Diagnostics

| Signal         | Interpretation                    |
| -------------- | --------------------------------- |
| Loss plateau   | Learning rate too high or too low |
| Exploding loss | Unstable gradients                |
| Overfitting    | Validation loss increasing        |
| Underfitting   | High training loss                |

---

### Conceptual Summary

The training loop is the **engine** of learning:

> Data → Prediction → Loss → Gradient → Parameter Update → Repeat

Its correctness, stability, and efficiency determine model quality, convergence speed, and generalization performance.
