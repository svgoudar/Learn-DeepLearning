
## Workflows


An **RBF Network** transforms your input data through **three key stages**:

1. **Input → Hidden Layer (RBF transformation)**: Convert raw inputs into localized activations using Gaussian functions.
2. **Hidden Layer → Output Layer (Linear combination)**: Combine these local activations linearly to produce final predictions.
3. **Training**: Learn where to place these Gaussian “centers,” how wide they should be, and what weights to assign at the output.

---

### Step-by-Step Workflow

#### **1. Input Data Preparation**

You start with input features $X = [x_1, x_2, ..., x_n]$.

Example:
If you’re approximating $y = \sin(x)$, your inputs might be numbers between 0 and $2\pi$.

No special normalization rules — but scaling helps if feature magnitudes differ greatly.

---

#### **2. Choose Number of RBF Neurons (Hidden Units)**

You decide how many **RBF centers** you need.
Each center represents a local region of influence in input space.

🧠 **Analogy**:
Each hidden neuron is a “local expert” responsible for a small neighborhood of the data.

Typical choices:

* Small dataset → few RBFs (5–10)
* Complex dataset → more RBFs (50+)

---

#### **3. Determine the Centers ($c_i$)**

Each RBF neuron has a **center** $c_i$, the location it is most responsive to.

Common ways to find centers:

* **K-Means Clustering**: Find representative points from your dataset.
* **Random Sampling**: Pick random points as centers.
* **Learnable Parameters**: Optimize $c_i$ during training (more costly).

📍 Each input $x$ is compared to these centers — distances decide activation strength.

---

#### **4. Compute the Spread (Width) $\sigma_i$**

The **spread** (or bandwidth) controls how wide each Gaussian bump is:
$$
\phi_i(x) = \exp\left(-\frac{|x - c_i|^2}{2\sigma_i^2}\right)
$$

* Small $\sigma_i$ → narrow bump (highly localized)
* Large $\sigma_i$ → wide bump (global influence)

How to set:

* Use **average distance** between neighboring centers
* Or learn $\sigma_i$ during training

---

#### **5. Transform Input → Hidden Layer**

Now, for each input $x$, compute activations from all RBF neurons:
$$
h_i(x) = \phi_i(x) = \exp\left(-\frac{|x - c_i|^2}{2\sigma_i^2}\right)
$$
This produces a vector of activations:
$$
H(x) = $$h_1(x), h_2(x), ..., h_M(x)$$
$$

🧠 **Intuition**:
Each neuron tells “how close” the input is to its region.
Far inputs get weak activations; nearby inputs get strong activations.

---

#### **6. Combine Hidden Activations in Output Layer**

The output layer performs a **weighted sum**:
$$
\hat{y} = W^T H(x)
$$
where $W$ are weights learned from data.

If the task is classification, a **softmax** or **sigmoid** may follow.
If regression, it’s just a linear sum.

---

#### **7. Train the Network**

Training has two main phases:

##### (a) **Fix the RBF parameters** (centers $c_i$, spreads $\sigma_i$)

Usually found using unsupervised learning (e.g., K-Means).

##### (b) **Optimize output weights $W$**

Once RBF activations are computed, the problem becomes **linear**:
$$
\hat{Y} = H W
$$
You can solve for $W$ using:

* **Least Squares Regression** (fast)
* Or **Gradient Descent** (for larger data)

🧮 The goal:
Minimize reconstruction/prediction error:
$$
\text{Loss} = |Y - \hat{Y}|^2
$$

---

#### **8. Prediction**

Once trained:

* Input a new $x$
* Compute its distance to all centers
* Get activations via Gaussian function
* Compute weighted sum → final prediction

---

### Intuitive Summary

| Stage           | What Happens                              | Analogy                                      |
| --------------- | ----------------------------------------- | -------------------------------------------- |
| 1. Input        | Feed raw data                             | Provide the problem                          |
| 2. Centers      | Define where each RBF neuron focuses      | Place “sensors” on a map                     |
| 3. Spread       | Define how far each neuron sees           | Each sensor’s range                          |
| 4. Hidden Layer | Compute how close input is to each sensor | Measure signal strength                      |
| 5. Output Layer | Combine signals with learned weights      | Fuse local opinions into one global decision |
| 6. Training     | Adjust centers, spreads, and weights      | Learn best sensor placement and weighting    |

---

### Mathematical Compact Form

$$
y(x) = \sum_{i=1}^{M} w_i , \exp\left(-\frac{|x - c_i|^2}{2\sigma_i^2}\right)
$$
where:

* $c_i$: center
* $\sigma_i$: width
* $w_i$: learned output weight

---

### Example Workflow (MNIST Digit Recognition)

1. Inputs: 784-pixel vectors.
2. Centers: Use K-Means to find 1000 cluster centers.
3. Hidden Layer: Compute Gaussian activations per input.
4. Output Layer: 10 neurons → digits (0–9).
5. Train output weights using softmax cross-entropy loss.

Result: A smooth, distance-based classifier that clusters similar digit shapes.
