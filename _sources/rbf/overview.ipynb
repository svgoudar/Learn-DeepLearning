Input (x) → [Hidden RBF Layer φ(||x−c||)] → Linear Output Layer
```

![alt text](../images/rbf.png)

---

### **Mathematical Intuition**

Each hidden neuron computes how *close* the input vector $x$ is to a *center* $c_i$.

$$
\phi(||x - c_i||) = e^{-\frac{||x - c_i||^2}{2\sigma_i^2}}
$$

Where:

* $x$: input vector
* $c_i$: center of the i-th RBF neuron
* $\sigma_i$: spread (controls how wide the Gaussian is)
* $||x - c_i||$: Euclidean distance between input and center

Each hidden neuron produces a **localized response** — high when input is near its center, low otherwise.

---

#### **4. Working Workflow**

1. **Input Layer:**
   Receives the feature vector $x$.

2. **Hidden Layer (RBF transformation):**
   Each neuron applies a **radial basis function** to measure similarity between $x$ and its center $c_i$.

   Example (Gaussian):
   $$
   \phi_i(x) = e^{-\frac{||x - c_i||^2}{2\sigma_i^2}}
   $$

3. **Output Layer (Linear combination):**
   Output is a linear combination of all RBF neuron activations:
   $$
   y(x) = \sum_{i=1}^{N} w_i \phi_i(x)
   $$
   where $w_i$ are learned weights.

---

#### **5. Training Process**

Training involves **two stages:**

1. **Unsupervised learning (hidden layer):**

   * Find the RBF centers $c_i$ (e.g., using *k-means clustering*).
   * Estimate spreads $\sigma_i$.

2. **Supervised learning (output layer):**

   * Compute weights $w_i$ via least squares regression.

This is faster and simpler than fully backpropagating all parameters like in standard neural networks.

---

#### **6. Intuition**

Think of RBF neurons as **landmarks** in input space:

* Each neuron activates strongly when an input is close to its landmark (center).
* The network “blends” these local responses to approximate any function.

So:

* **RBFs act locally**, unlike ReLU or sigmoid activations which are global.
* The network can model complex, nonlinear shapes efficiently.

---

#### **7. Example**

For a 1D function $y = \sin(x)$:

* Input: $x \in [-π, π]
* Hidden layer: Gaussian RBFs centered across x-range
* Output: Weighted sum of RBFs approximates the sine curve

RBFs form a **smooth curve** passing near training points — great for interpolation.

---

#### **Applications**

| Domain                      | Use Case                                         |
| --------------------------- | ------------------------------------------------ |
| **Function approximation**  | Predict nonlinear functions                      |
| **Classification**          | Pattern recognition (e.g., handwritten digits)   |
| **Time series forecasting** | Predict stock, temperature                       |
| **Control systems**         | Adaptive control, robotics                       |
| **Anomaly detection**       | Detect outliers by distance from learned centers |

---

**Advantages**

* Simple, interpretable structure
* Fast training (especially output layer)
* Good at modeling local nonlinearities
* Robust for small datasets

---

**Disadvantages**

* Choosing the number and position of centers can be tricky
* Poor scalability for high-dimensional data
* Sensitive to spread parameter ($\sigma$)

---

**Summary Table**

| Property    | RBF Network                        |
| ----------- | ---------------------------------- |
| Type        | Feedforward                        |
| Activation  | Gaussian or other radial basis     |
| Learning    | Hybrid (unsupervised + supervised) |
| Strength    | Local approximation                |
| Weakness    | High-dimensional inefficiency      |
| Typical Use | Regression, function approximation |


```{dropdown} Click here for Sections
```{tableofcontents}
